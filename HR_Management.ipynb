```python

# ==========================================
# CORPORATE HR MANAGEMENT SYSTEM
# STAGE 1: OBJECT-ORIENTED PROGRAMMING
# ==========================================

# Base Class
class Employee:

    def __init__(self, emp_id, name, salary):
        self.__emp_id = emp_id
        self.__name = name
        self.__salary = salary

    # Getter methods
    def get_emp_id(self):
        return self.__emp_id

    def get_name(self):
        return self.__name

    def get_salary(self):
        return self.__salary

    # Display employee details
    def display_details(self):
        print("\n--- Employee Details ---")
        print("Employee ID:", self.__emp_id)
        print("Name:", self.__name)
        print("Salary:", self.__salary)

    # Convert object into dictionary
    def to_dict(self):
        return {
            "Employee ID": self.__emp_id,
            "Name": self.__name,
            "Salary": self.__salary,
            "Role": "Employee"
        }


# Derived Class: Manager
class Manager(Employee):

    def __init__(self, emp_id, name, salary,
                 department, team_size):

        super().__init__(emp_id, name, salary)

        self.department = department
        self.team_size = team_size

    # Method Overriding
    def display_details(self):
        print("\n--- Manager Details ---")
        print("Employee ID:", self.get_emp_id())
        print("Name:", self.get_name())
        print("Salary:", self.get_salary())
        print("Department:", self.department)
        print("Team Size:", self.team_size)

    def to_dict(self):
        data = super().to_dict()
        data.update({
            "Role": "Manager",
            "Department": self.department,
            "Team Size": self.team_size
        })
        return data


# Derived Class: Intern
class Intern(Employee):

    def __init__(self, emp_id, name, salary,
                 college, duration):

        super().__init__(emp_id, name, salary)

        self.college = college
        self.duration = duration

    # Method Overriding
    def display_details(self):
        print("\n--- Intern Details ---")
        print("Employee ID:", self.get_emp_id())
        print("Name:", self.get_name())
        print("Stipend:", self.get_salary())
        print("College:", self.college)
        print("Duration:", self.duration)

    def to_dict(self):
        data = super().to_dict()
        data.update({
            "Role": "Intern",
            "College": self.college,
            "Duration": self.duration
        })
        return data


# ==========================================
# TESTING THE CLASSES
# ==========================================

employees = []

# Creating Employee object
e1 = Employee("E101", "Rahul", 30000)

# Creating Manager object
m1 = Manager("M101", "Amit", 60000,
             "IT", 10)

# Creating Intern object
i1 = Intern("I101", "Priya", 12000,
            "Brainware University", "3 Months")

# Store all objects in one list
employees.append(e1)
employees.append(m1)
employees.append(i1)

# Display all employee records
for emp in employees:
    emp.display_details()

# Display dictionary representation
print("\n--- Dictionary Data ---")
for emp in employees:
    print(emp.to_dict())
```

**CSV File Handling**

```python

import csv
import os

# CSV file name
FILE_NAME = "employees.csv"

# Column names
FIELDS = [
    "Employee ID",
    "Name",
    "Salary",
    "Role",
    "Department",
    "Team Size",
    "College",
    "Duration"
]


# 1. SAVE EMPLOYEE DATA INTO CSV
def save_to_csv(employees):

    with open(FILE_NAME, "w", newline="", encoding="utf-8") as file:

        writer = csv.DictWriter(file, fieldnames=FIELDS)

        writer.writeheader()

        for emp in employees:
            writer.writerow(emp.to_dict())

    print("Employee records saved successfully!")


# 2. READ DATA FROM CSV
def read_from_csv():

    if not os.path.exists(FILE_NAME):
        print("CSV file does not exist!")
        return []

    with open(FILE_NAME, "r", encoding="utf-8") as file:

        reader = csv.DictReader(file)

        records = list(reader)

    return records


# 3. DISPLAY CSV RECORDS
def display_csv():

    records = read_from_csv()

    if not records:
        print("No records found!")
        return

    print("\n===== EMPLOYEE RECORDS =====")

    for record in records:
        print("\nEmployee ID:", record["Employee ID"])
        print("Name:", record["Name"])
        print("Salary:", record["Salary"])
        print("Role:", record["Role"])


# 4. SEARCH EMPLOYEE BY ID
def search_employee(emp_id):

    records = read_from_csv()

    for record in records:

        if record["Employee ID"] == emp_id:
            print("\nEmployee Found!")
            print(record)
            return

    print("Employee not found!")


# 5. ADD A NEW RECORD TO CSV
def add_to_csv(emp):

    records = read_from_csv()

    records.append(emp.to_dict())

    with open(FILE_NAME, "w", newline="", encoding="utf-8") as file:

        writer = csv.DictWriter(file, fieldnames=FIELDS)

        writer.writeheader()
        writer.writerows(records)

    print("New employee added successfully!")


# ===================================
# TESTING CSV OPERATIONS
# ===================================

# Save Stage 1 employee objects
save_to_csv(employees)

# Display all records
display_csv()

# Search employee
search_employee("E101")

# Create a new employee
e2 = Employee("E102", "Rohit", 35000)

# Add new employee to CSV
add_to_csv(e2)

# Display updated records
display_csv()
```

JSON file **handling**

```python

import json
import os

# JSON file name
JSON_FILE = "employees.json"


# 1. SAVE EMPLOYEE DATA INTO JSON
def save_to_json(employees):

    data = []

    for emp in employees:
        data.append(emp.to_dict())

    with open(JSON_FILE, "w", encoding="utf-8") as file:

        json.dump(data, file, indent=4)

    print("Employee records saved in JSON successfully!")


# 2. READ DATA FROM JSON
def read_from_json():

    if not os.path.exists(JSON_FILE):
        print("JSON file does not exist!")
        return []

    try:
        with open(JSON_FILE, "r", encoding="utf-8") as file:
            data = json.load(file)

        return data

    except json.JSONDecodeError:
        print("Invalid JSON file!")
        return []


# 3. DISPLAY ALL JSON RECORDS
def display_json():

    records = read_from_json()

    if not records:
        print("No records found!")
        return

    print("\n===== JSON EMPLOYEE RECORDS =====")

    for record in records:

        print("\nEmployee ID:", record.get("Employee ID"))
        print("Name:", record.get("Name"))
        print("Salary:", record.get("Salary"))
        print("Role:", record.get("Role"))

        if record.get("Role") == "Manager":
            print("Department:", record.get("Department"))
            print("Team Size:", record.get("Team Size"))

        elif record.get("Role") == "Intern":
            print("College:", record.get("College"))
            print("Duration:", record.get("Duration"))


# 4. SEARCH EMPLOYEE BY ID
def search_json(emp_id):

    records = read_from_json()

    for record in records:

        if record.get("Employee ID") == emp_id:
            print("\nEmployee Found!")
            print(json.dumps(record, indent=4))
            return

    print("Employee not found!")


# 5. ADD A NEW EMPLOYEE TO JSON
def add_to_json(emp):

    records = read_from_json()

    # Check for duplicate Employee ID
    for record in records:

        if record.get("Employee ID") == emp.get_emp_id():
            print("Employee ID already exists!")
            return

    records.append(emp.to_dict())

    with open(JSON_FILE, "w", encoding="utf-8") as file:

        json.dump(records, file, indent=4)

    print("New employee added successfully!")


# 6. CONVERT JSON DATA INTO PYTHON OBJECTS
def load_json_objects():

    records = read_from_json()
    objects = []

    for record in records:

        role = record.get("Role")

        if role == "Manager":

            emp = Manager(
                record["Employee ID"],
                record["Name"],
                float(record["Salary"]),
                record["Department"],
                int(record["Team Size"])
            )

        elif role == "Intern":

            emp = Intern(
                record["Employee ID"],
                record["Name"],
                float(record["Salary"]),
                record["College"],
                record["Duration"]
            )

        else:

            emp = Employee(
                record["Employee ID"],
                record["Name"],
                float(record["Salary"])
            )

        objects.append(emp)

    return objects


# ==================================
# TESTING JSON OPERATIONS
# ==================================

# Save existing OOP objects
save_to_json(employees)

# Display JSON records
display_json()

# Search employee
search_json("E101")

# Add a new manager
m2 = Manager(
    "M102", "Ankit", 70000,
    "HR", 8
)

add_to_json(m2)

# Display updated records
display_json()

# Load JSON data into Python objects
loaded_employees = load_json_objects()

print("\n===== LOADED PYTHON OBJECTS =====")

for emp in loaded_employees:
    emp.display_details()
```

```python

import pandas as pd
import os

EXCEL_FILE = "employees.xlsx"


# 1. SAVE EMPLOYEE DATA INTO EXCEL
def save_to_excel(employees):

    data = []

    for emp in employees:
        data.append(emp.to_dict())

    df = pd.DataFrame(data)

    df.to_excel(EXCEL_FILE, index=False)

    print("Employee records saved to Excel successfully!")


# 2. READ EMPLOYEE DATA FROM EXCEL
def read_from_excel():

    if not os.path.exists(EXCEL_FILE):
        print("Excel file does not exist!")
        return pd.DataFrame()

    df = pd.read_excel(EXCEL_FILE)

    return df


# 3. DISPLAY ALL EMPLOYEE RECORDS
def display_excel():

    df = read_from_excel()

    if df.empty:
        print("No employee records found!")
    else:
        print("\n===== EMPLOYEE RECORDS =====")
        display(df)


# 4. SEARCH EMPLOYEE BY ID
def search_excel(emp_id):

    df = read_from_excel()

    if df.empty:
        print("No records available!")
        return

    result = df[df["Employee ID"] == emp_id]

    if result.empty:
        print("Employee not found!")
    else:
        print("\nEmployee Found!")
        display(result)


# 5. FILTER EMPLOYEES BY ROLE
def filter_by_role(role):

    df = read_from_excel()

    if df.empty:
        print("No records available!")
        return

    result = df[df["Role"].str.lower() == role.lower()]

    print("\n===== FILTERED EMPLOYEES =====")
    display(result)


# 6. CALCULATE TOTAL AND AVERAGE SALARY
def salary_analysis():

    df = read_from_excel()

    if df.empty:
        print("No records available!")
        return

    total = df["Salary"].sum()
    average = df["Salary"].mean()

    print("\nTotal Salary:", total)
    print("Average Salary:", average)


# 7. ADD NEW EMPLOYEE RECORD
def add_excel_record(emp):

    df = read_from_excel()

    if not df.empty:
        if emp.get_emp_id() in df["Employee ID"].astype(str).values:
            print("Employee ID already exists!")
            return

    new_record = pd.DataFrame([emp.to_dict()])

    df = pd.concat([df, new_record], ignore_index=True)

    df.to_excel(EXCEL_FILE, index=False)

    print("New employee added successfully!")


# =====================================
# TESTING EXCEL OPERATIONS
# =====================================

# Save existing employees
save_to_excel(employees)

# Display records
display_excel()

# Search employee
search_excel("E101")

# Filter managers
filter_by_role("Manager")

# Salary analysis
salary_analysis()

# Add new employee
e3 = Employee("E103", "Arjun", 40000)

add_excel_record(e3)

# Display updated records
display_excel()
```

```python

import gradio as gr
import pandas as pd
import csv
import json
import os

CSV_FILE = "employees.csv"
JSON_FILE = "employees.json"
EXCEL_FILE = "employees.xlsx"

FIELDS = [
    "Employee ID", "Name", "Salary", "Role",
    "Department", "Team Size", "College", "Duration"
]

# ---------------- OOP CLASSES ----------------

class Employee:
    def __init__(self, emp_id, name, salary):
        self.__emp_id = emp_id
        self.__name = name
        self.__salary = float(salary)

    def get_id(self):
        return self.__emp_id

    def get_name(self):
        return self.__name

    def get_salary(self):
        return self.__salary

    def display_details(self):
        return f"Employee: {self.__name}, Salary: {self.__salary}"

    def to_dict(self):
        return {
            "Employee ID": self.__emp_id,
            "Name": self.__name,
            "Salary": self.__salary,
            "Role": "Employee",
            "Department": "",
            "Team Size": "",
            "College": "",
            "Duration": ""
        }


class Manager(Employee):
    def __init__(self, emp_id, name, salary, department, team_size):
        super().__init__(emp_id, name, salary)
        self.department = department
        self.team_size = int(team_size)

    def display_details(self):
        return f"Manager: {self.get_name()}, Department: {self.department}"

    def to_dict(self):
        data = super().to_dict()
        data.update({
            "Role": "Manager",
            "Department": self.department,
            "Team Size": self.team_size
        })
        return data


class Intern(Employee):
    def __init__(self, emp_id, name, salary, college, duration):
        super().__init__(emp_id, name, salary)
        self.college = college
        self.duration = duration

    def display_details(self):
        return f"Intern: {self.get_name()}, College: {self.college}"

    def to_dict(self):
        data = super().to_dict()
        data.update({
            "Role": "Intern",
            "College": self.college,
            "Duration": self.duration
        })
        return data


# ---------------- DATA STORAGE ----------------

def load_data():
    if os.path.exists(CSV_FILE):
        return pd.read_csv(CSV_FILE).fillna("")
    return pd.DataFrame(columns=FIELDS)


def save_data(df):
    df.to_csv(CSV_FILE, index=False)
    df.to_json(JSON_FILE, orient="records", indent=4)
    df.to_excel(EXCEL_FILE, index=False)


# ---------------- GUI FUNCTIONS ----------------

def add_employee(emp_id, name, salary, role,
                 department, team_size, college, duration):

    try:
        if not emp_id or not name or not salary or not role:
            return "Please fill in all required fields.", load_data()

        salary = float(salary)

        if salary < 0:
            return "Salary cannot be negative.", load_data()

        df = load_data()

        if emp_id in df["Employee ID"].astype(str).values:
            return "Employee ID already exists.", df

        if role == "Manager":
            if not department or not str(team_size).strip():
                return "Enter department and team size.", df

            obj = Manager(emp_id, name, salary, department, team_size)

        elif role == "Intern":
            if not college or not duration:
                return "Enter college and duration.", df

            obj = Intern(emp_id, name, salary, college, duration)

        else:
            obj = Employee(emp_id, name, salary)

        new_record = pd.DataFrame([obj.to_dict()])
        df = pd.concat([df, new_record], ignore_index=True)

        save_data(df)

        return "Employee added successfully!", df

    except Exception as e:
        return f"Error: {e}", load_data()


def search_employee(emp_id):
    df = load_data()
    result = df[df["Employee ID"].astype(str) == str(emp_id)]

    if result.empty:
        return "Employee not found.", result

    return "Employee found!", result


def delete_employee(emp_id):
    df = load_data()
    updated = df[df["Employee ID"].astype(str) != str(emp_id)]

    if len(updated) == len(df):
        return "Employee ID not found.", df

    save_data(updated)
    return "Employee deleted successfully!", updated


def show_all():
    return load_data()


def export_message():
    df = load_data()
    save_data(df)
    return "CSV, JSON and Excel files are ready for download."


# ---------------- GRADIO INTERFACE ----------------

with gr.Blocks(title="Corporate HR Management System") as app:

    gr.Markdown("# Corporate HR Management System")
    gr.Markdown("Manage Employees, Managers and Interns using Python OOP.")

    with gr.Tab("Add Employee"):
        emp_id = gr.Textbox(label="Employee ID")
        name = gr.Textbox(label="Employee Name")
        salary = gr.Number(label="Salary")
        role = gr.Dropdown(
            ["Employee", "Manager", "Intern"],
            label="Role"
        )

        department = gr.Textbox(label="Department (Manager)")
        team_size = gr.Number(label="Team Size (Manager)")
        college = gr.Textbox(label="College (Intern)")
        duration = gr.Textbox(label="Internship Duration")

        add_btn = gr.Button("Add Employee")
        add_status = gr.Textbox(label="Status")
        add_table = gr.Dataframe(value=load_data(), label="Employee Records")

        add_btn.click(
            add_employee,
            inputs=[
                emp_id, name, salary, role,
                department, team_size, college, duration
            ],
            outputs=[add_status, add_table]
        )

    with gr.Tab("Search Employee"):
        search_id = gr.Textbox(label="Enter Employee ID")
        search_btn = gr.Button("Search")
        search_status = gr.Textbox(label="Status")
        search_table = gr.Dataframe()

        search_btn.click(
            search_employee,
            inputs=search_id,
            outputs=[search_status, search_table]
        )

    with gr.Tab("Delete Employee"):
        delete_id = gr.Textbox(label="Enter Employee ID")
        delete_btn = gr.Button("Delete")
        delete_status = gr.Textbox(label="Status")
        delete_table = gr.Dataframe()

        delete_btn.click(
            delete_employee,
            inputs=delete_id,
            outputs=[delete_status, delete_table]
        )

    with gr.Tab("View All Employees"):
        view_btn = gr.Button("Refresh Records")
        view_table = gr.Dataframe(value=load_data())

        view_btn.click(show_all, outputs=view_table)

    with gr.Tab("Export Files"):
        export_btn = gr.Button("Prepare CSV, JSON and Excel")
        export_status = gr.Textbox()

        export_btn.click(export_message, outputs=export_status)

        gr.Markdown(
            "After preparing the files, download them from "
            "the Colab Files panel on the left."
        )

app.launch(share=True, debug=True)
```
